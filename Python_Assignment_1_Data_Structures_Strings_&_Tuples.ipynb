# String Slicing and Indexing:

```python
""" a) In this program, we use indexing and slicing Python starts indexing from
 0,so result[0] gives the first character"""


result="Hello Zara welcome to Python programming"
print(result[0])
```

```python
""" b) In Python, negative indexing starts from the end.so result[-1] gives the
last character"""


print(result[-1:])
```

```python
""" c)Start from the beginning and stop before index 5.Python does not include
the ending index.only indexes 0, 1, 2, 3, 4 are taken."""

print(result[:5])
```

```python
""" d)-11 means start 11 characters from the end.which means it Takes the last
11 characters."""

print(result[-11:])
```

```python
""" e) [::-1] means start from the beginning,go to the end,move with a step of-1
A step of -1 means move backwards. """

print(result[::-1])
```

```python
""" f) We use slicing to extract a particular part of the string.Start at index
-18 and stop before index -11"""

print(result[-18:-11])
```

# String Methods:

```python
""" a) by using .upper() we can converts all letters into uppercase.
it is a string method."""


strM ="Python beginner tutorial"
print(strM.upper())
```

```python
""" b) by using .lower() we can converts all letters into lowercase.
it is a string method."""


print(strM.lower())
```

```python
""" c) .capitalize() makes the first character uppercase and converts the
remaining letters to lowercase."""


print(strM.capitalize())
```

```python
""" d).count() counts how many times something appears in a string. """

print(strM.count("t"))
```

```python
""" e) .replace() is used when we want to replace one piece of text with
another. """

print(strM.replace("Python", "Machine Learning"))
```

```python
""" a) A tuple is a collection of values written inside parentheses ().
we create two tuples containing three numbers each. We use the + operator to
concatenate the two tuples and store the result in t_combine"""

a=(10,20,30)
b=(40,50,60)

print(a)
print(b)
t_combine= a + b
print (t_combine)
```

```python
""" b)We use * 3 to repeat the combined tuple three times"""

print(t_combine * 3)
```

```python
""" c)Python indexing starts from 0, t_combine[2] gives the third element """

print(t_combine[2])
```

```python
""" d) [:3] to get the first three elements"""

print(t_combine[:3])
```

```python
""" e) [-3:] to get the last three elements."""

print(t_combine[-3:])
```
